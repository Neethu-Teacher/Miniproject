## Account details

```python
account__number="016105300062004"
pin="3006"
balance=5000.0
print("====BANKING SYSTEM====")
```

# Account log in

```python
username = input("Enter account number/username:")
password = input("Enter PIN/password:")
if username == account__number and password == pin:
  print("\nLogin successful!")
  while True:
    print("\n----Menu----")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
      amount = float(input("Enter deposit amount="))
      if amount > 0:
        balance += amount
        print(f"Deposit successful.")
        print(f"Current balance: {balance:.2f}")
      else:
        print("Deposit amount must be greater than zero.")
    elif choice == "2":
      amount = float(input("Enter withdrawal amount:"))
      if amount <= 0:
        print("Withdrawal amount must be greater than zero.")
      elif amount > balance:
        print("Insufficient balance. Withdrawal amount cannot exceed available balance.")
      else:
        balance -= amount
        print("Withdrawal successful.")
        print(f"Remaining balance:₹{balance:.2f}")
    elif choice == "3":
      print(f"Your current balance is:₹{balance:.2f}")
    elif choice == "4":
      print("Exiting banking system. Goodbye!")
      break
    else:
      print("Invalid choice. Please select a valid option (1-4).")
else:
  print("Invalid account number or PIN. Please try again.")
```
